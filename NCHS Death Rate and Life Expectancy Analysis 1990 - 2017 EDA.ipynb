```python
# Individual variable istribution and characteristics
sns.histplot(df['Average Life Expectancy (Years)'])
plt.title('Plot of Average Life Expectancy (Years)')
plt.xlabel('Average Life Expectancy (Years)')
plt.ylabel('Count')
plt.show()
```

```python
sns.boxplot(df['Average Life Expectancy (Years)'])  
plt.show()
```

```python
sns.histplot(df['Age-adjusted Death Rate'])
plt.title('Plot of Age-adjusted Death Rate')
plt.xlabel('Age-adjusted Death Rate')
plt.ylabel('Count')
plt.show()
```

```python
sns.boxplot(df['Age-adjusted Death Rate'])  
plt.show()  
```

```python
df['Race'].value_counts().plot(kind='bar')
plt.show()
```

```python
df['Sex'].value_counts().plot(kind='bar')
plt.show()
```

```python
# Relationship brtween variables
sns.scatterplot(x= 'Average Life Expectancy (Years)', y = 'Year', hue='Sex', data = df)
plt.title('Scatterplot of Average Life Expectancy (Years) and Year')
plt.xlabel('Average Life Expectancy (Years)')
plt.ylabel('Year')
plt.show()
```

```python
# Live expectancy and Death Rate correlation 
sns.scatterplot(x='Average Life Expectancy (Years)', y='Age-adjusted Death Rate', data=df)  
plt.show()
```

```python
# Correlation Matrix
corr = df[['Average Life Expectancy (Years)', 'Age-adjusted Death Rate']].corr()
corr
```

```python
corr = df[['Average Life Expectancy (Years)', 'Age-adjusted Death Rate']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

```python
# Average Life Expectancy vs Race
sns.boxplot(x='Race', y='Average Life Expectancy (Years)', data=df)
plt.show()
```

```python
# Average Life Expectancy vs Gender
sns.boxplot(x='Sex', y='Average Life Expectancy (Years)', data=df)
plt.show()
```
